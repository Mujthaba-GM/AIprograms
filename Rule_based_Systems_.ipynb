```python
# Medical Diagnosis (Simple Expert System)
fever = input("Do you have fever (yes/no): ")
cough = input("Do you have cough (yes/no): ")

if fever == "yes" and cough == "yes":
    print("Possible Flu")
elif fever == "yes":
    print("Possible Viral Fever")
elif cough == "yes":
    print("Common Cold")
else:
    print("You seem Healthy")
```

```python
#Traffic Signal Decision
color = input("Enter Signal Color (red/yellow/green): ")

if color == "red":
    print("STOP")
elif color == "yellow":
    print("READY")
elif color == "green":
    print("GO")
else:
    print("Invalid Signal")
```

```python
#Weather Recommendation
weather = input("Enter weather (sunny/rainy/cold): ")

if weather == "sunny":
    print("Wear Light Clothes")
elif weather == "rainy":
    print("Carry an Umbrella")
elif weather == "cold":
    print("Wear a Jacket")
else:
    print("Weather Unknown")
```

```python
# Loan Approval System
salary = int(input("Monthly Salary: "))
age = int(input("Age: "))

if salary >= 30000 and age >= 21:
    print("Loan Approved")
else:
    print("Loan Rejected")
```

```python
#Rule based system medical diagnosis systems
fever = input("Do you have fever? (yes/no): ").lower()
cough = input("Do you have cough? (yes/no): ").lower()
headache = input("Do you have headache? (yes/no): ").lower()
bodypain = input("Do you have body pain? (yes/no): ").lower()
breathing = input("Do you have breathing difficulty? (yes/no): ").lower()
temperature = float(input("Enter body temperature (°F): "))

print("\nDiagnosis Result")

# Rule 1
if fever == "yes" and cough == "yes" and breathing == "yes":
    print("Disease : Pneumonia")
    print("Priority : Emergency")

# Rule 2
elif fever == "yes" and bodypain == "yes" and headache == "yes":
    print("Disease : Viral Fever")
    print("Advice : Take Rest and Drink Plenty of Fluids")

# Rule 3
elif fever == "yes" and cough == "yes":
    print("Disease : Common Cold")

# Rule 4
elif temperature > 103:
    print("High Fever")
    print("Hospital Admission Recommended")

# Rule 5
elif breathing == "yes":
    print("Consult a Pulmonologist Immediately")

# Rule 6
elif fever == "yes":
    print("Possible Mild Infection")

# Rule 7
else:
    print("No Serious Symptoms Detected")
    print("You appear to be Healthy")
```
