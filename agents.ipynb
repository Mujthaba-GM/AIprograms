```python
# Automatic door (Simple Reflex agents)
def automatic_door(person):
    if person == "yes":
        return "Open Door"
    else:
        return "Close Door"


print(automatic_door("yes"))
print(automatic_door("no"))
```

```python
#Use Case 2: Thermostat (Simple Reflex agents)
def thermostat(temp):
    if temp < 20:
        return "Heater ON"
    elif temp > 25:
        return "Heater OFF"
    else:
        return "Normal Temperature"


print(thermostat(18))
print(thermostat(23))
print(thermostat(28))
```

```python
#Use Case 1: Robot Vacuum Cleaner (Model-Based Reflex Agent)
def robot_vacuum(room, cleaned):

    if room in cleaned:
        return "Room already cleaned"
    else:
        cleaned.append(room)
        return "Cleaning " + room


cleaned = []

print(robot_vacuum("Room A", cleaned))
print(robot_vacuum("Room B", cleaned))
print(robot_vacuum("Room A", cleaned))

print("Memory:", cleaned)
```

```python
#Use Case 2: Robot with Obstacle Memory (Model-Based Reflex Agent)
def robot_move(direction, obstacles):

    if direction in obstacles:
        return "Obstacle detected - Stop"
    else:
        return "Move " + direction


obstacles = ["left"]

print(robot_move("left", obstacles))
print(robot_move("right", obstacles))
```

```python
# GPS navigation (Goal-Based Agent)
def navigation(current, destination):

    if current == destination:
        return "Destination Reached"
    else:
        return "Move towards " + destination


print(navigation("Hyderabad", "Vijayawada"))
print(navigation("Vijayawada", "Hyderabad"))
```

```python
#GPS navigation (Goal-Based Agent) Use Case 2: Delivery Robot
def delivery_robot(location, goal):

    if location == goal:
        return "Package Delivered"
    else:
        return "Move towards " + goal


print(delivery_robot("Room 101", "Room 205"))
print(delivery_robot("Room 205", "Room 205"))
```

```python
#Utility-Based Agent Use Case 1: Route Selection
routes = {
    "Route A": 70,
    "Route B": 90,
    "Route C": 60
}

best_route = max(routes, key=routes.get)

print("Best Route:", best_route)
print("Utility:", routes[best_route])
```

```python
#Use Case 2: Self-Driving Car
routes = {
    "Route A": 80,
    "Route B": 60,
    "Route C": 95
}

best_route = max(routes, key=routes.get)

print("Selected Route:", best_route)
print("Safety Utility:", routes[best_route])
```

```python
#Learning Agent Use Case 1: Email Spam Filter
spam_words = ["win", "prize", "offer", "free"]

def spam_filter(email):

    for word in spam_words:
        if word in email.lower():
            return "Spam"

    return "Not Spam"


print(spam_filter("You won a free prize"))
print(spam_filter("Tomorrow is a meeting"))
```

```python
#Learning Agent Use Case 2: Recommendation System
def recommendation(history):

    if "cricket" in history:
        return "Recommend Cricket Videos"

    elif "football" in history:
        return "Recommend Football Videos"

    else:
        return "Recommend Popular Videos"


history = ["cricket", "cricket", "movies"]

print(recommendation(history))
```
