```python
# BFS using Queue

graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['G', 'H'],
    'D': ['F'],
    'E': ['F'],
    'G': ['I'],
    'H': ['I'],
    'F': ['J'],
    'I': ['J'],
    'J': []
}

queue = []
visited = []

# Start with A
queue.append('A')
visited.append('A')

while queue:
    x = queue.pop(0)
    print("Delete:", x)

    for node in graph[x]:
        if node not in visited:
            visited.append(node)
            queue.append(node)
            print("Visit:", node)
            print("Queue:", queue)

print("\nBFS Traversal:")
print(" -> ".join(visited))
```
